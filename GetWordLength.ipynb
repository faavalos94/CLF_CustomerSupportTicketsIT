### Use for your sample data: Replace with any text to get word length

```python
from nltk import word_tokenize
```

```python
# Example texts to show how word_tokenize works
texts = ['How are you doing?', "I'm doing good.", 'What did you have for lunch today?']
```

```python
def get_word_len(texts):
    word_tokens = [word_tokenize(text) for text in texts]

    len_tokens = []

    for i in range(len(word_tokens)):
        len_tokens.append(len(word_tokens[i]))

    return len_tokens
```

```python
print(get_word_len(texts))
```
